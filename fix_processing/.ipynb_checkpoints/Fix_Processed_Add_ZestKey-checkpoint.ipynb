# Fix processed_* — re-attach ZEST_KEY

The PrepareData phase-2 save did `agg_eng.transform(sub).to_parquet(path, index=False)`. The aggregator
returns one row per applicant **indexed by ZEST_KEY**, and `index=False` dropped that index — so every
`processed_new`/`processed_old` file kept the 8,848 `trade_*` features but **lost ZEST_KEY**, leaving
nothing to join trade -> app/target on.

**Why this is a clean, local fix:** the NEW-vs-OLD aggregator difference lives entirely in
`normalized_*` (phase-1 `PreprocessorV2`, which runs `PaymentPatternsAggregatorV2`). Phase-2's
`AggregationEngine` (`aggregation/fe2/trade.json`) is the **same** for both variants, and
`normalized_*` still carries `ZEST_KEY`. So we just re-run phase 2 from `normalized_<variant>` and
save the key. No Snowflake, no re-deciding new/old.

For each `(bureau, split, variant)` we ran:
1. **back up** the current keyless `processed_<variant>/` -> `processed_<variant>_without_zest_key/`
2. **regenerate** `processed_<variant>/` from `normalized_<variant>` (same buckets, same aggregator),
   saving with `ZEST_KEY` as a column.

**Kernel:** run this in the same model-engine kernel the PrepareData notebooks used (it imports
`model_engine`). Re-aggregation is the slow part, but it's local CPU only.

```python
import os, sys, gc, glob, shutil, warnings
import pandas as pd

# configs.py lives one level up (payment-processor-research/)
for cand in [os.path.abspath(os.path.join(os.getcwd(), '..')), os.getcwd(),
             '/home/jag/payment-processor-research']:
    if os.path.exists(os.path.join(cand, 'configs.py')):
        sys.path.insert(0, cand); break
import configs
from configs import DATA_DIR

from model_engine.assets.utils import load_asset
from model_engine.feature_engine_V2.feature_engine import AggregationEngine
warnings.filterwarnings('ignore')

N_BUCKETS = 100   # must match what PrepareData used (part-000..099)

# Same aggregator for every bureau/split/variant (it works on standardized normalized columns).
agg_eng = AggregationEngine(asset=load_asset('aggregation/fe2/trade.json'), table_name='trade')
print('AggregationEngine ready  |  DATA_DIR =', DATA_DIR)
```

```python
# Discover every (bureau, split, variant) that actually has a processed_* dir with parts.
combos = []
for variant in ['new', 'old']:
    for d in sorted(glob.glob(os.path.join(DATA_DIR, '*', '*', f'processed_{variant}'))):
        if not glob.glob(os.path.join(d, 'part-*.parquet')):
            continue
        split  = os.path.basename(os.path.dirname(d))
        bureau = os.path.basename(os.path.dirname(os.path.dirname(d)))
        combos.append((bureau, split, variant))

for b, s, v in combos:
    n = len(glob.glob(os.path.join(DATA_DIR, b, s, f'processed_{v}', 'part-*.parquet')))
    print(f'{b:<11} {s:<6} {v:<4} {n:>4} parts')
print(f'\n{len(combos)} combos to fix')
```

```python
def fix_one(bureau, split, variant):
    proc_dir = os.path.join(DATA_DIR, bureau, split, f'processed_{variant}')
    backup   = os.path.join(DATA_DIR, bureau, split, f'processed_{variant}_without_zest_key')
    norm_dir = os.path.join(DATA_DIR, bureau, split, f'normalized_{variant}')

    if not glob.glob(os.path.join(norm_dir, 'part-*.parquet')):
        print(f'  SKIP: no normalized_{variant} at {norm_dir}'); return

    # 1) back up the keyless processed (only the FIRST time, so we never clobber it with a keyed run)
    if glob.glob(os.path.join(backup, 'part-*.parquet')):
        print(f'  backup already exists -> {backup}')
    elif os.path.isdir(proc_dir):
        shutil.move(proc_dir, backup)
        print(f'  backed up keyless -> {backup}')
    os.makedirs(proc_dir, exist_ok=True)
    for old in glob.glob(os.path.join(proc_dir, 'part-*.parquet')):
        os.remove(old)

    # 2) re-run phase 2 from normalized, saving ZEST_KEY (identical buckets + aggregator)
    df = pd.read_parquet(norm_dir)
    df['_bucket'] = (pd.util.hash_pandas_object(df['ZEST_KEY'], index=False) % N_BUCKETS).astype('int16')
    total = 0
    for b in range(N_BUCKETS):
        sub = df[df['_bucket'] == b].drop(columns='_bucket')
        if not len(sub):
            continue
        processed = agg_eng.transform(sub).reset_index()       # ZEST_KEY index -> column
        if 'ZEST_KEY' not in processed.columns and 'index' in processed.columns:
            processed = processed.rename(columns={'index': 'ZEST_KEY'})
        processed.to_parquet(os.path.join(proc_dir, f'part-{b:03d}.parquet'), index=False)
        total += len(processed)
        del sub, processed; gc.collect()
    del df; gc.collect()

    chk = pd.read_parquet(os.path.join(proc_dir, 'part-000.parquet'))
    assert 'ZEST_KEY' in chk.columns, f'ZEST_KEY STILL missing for {bureau}/{split}/{variant}!'
    print(f'  done: {total:,} rows, ZEST_KEY present ({chk.shape[1]} cols)')


for bureau, split, variant in combos:
    print(f'=== {bureau}/{split}/{variant} ===')
    fix_one(bureau, split, variant)
print('\nALL DONE')
```

```python
# Confirm every regenerated processed_* now has ZEST_KEY, and the keyless backups exist.
import pyarrow.parquet as pq
for bureau, split, variant in combos:
    proc   = os.path.join(DATA_DIR, bureau, split, f'processed_{variant}', 'part-000.parquet')
    backup = os.path.join(DATA_DIR, bureau, split, f'processed_{variant}_without_zest_key')
    has_key = os.path.exists(proc) and 'ZEST_KEY' in pq.read_schema(proc).names
    has_bak = bool(glob.glob(os.path.join(backup, 'part-*.parquet')))
    print(f'{bureau:<11} {split:<6} {variant:<4} | ZEST_KEY={has_key} | backup={has_bak}')
```
