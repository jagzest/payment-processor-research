# Fix processed — equifax / test / new

Backs up the keyless `processed_new/` to `processed_new_without_zest_key/`, then regenerates it from `normalized_new/` with **ZEST_KEY** kept. Run in the model-engine kernel.

```python
import os, sys, gc, glob, shutil, warnings
import pandas as pd

for cand in [os.path.abspath(os.path.join(os.getcwd(), '..')), os.getcwd(),
             '/home/jag/payment-processor-research']:
    if os.path.exists(os.path.join(cand, 'configs.py')):
        sys.path.insert(0, cand); break
import configs
from configs import DATA_DIR

from model_engine.assets.utils import load_asset
from model_engine.feature_engine_V2.feature_engine import AggregationEngine
warnings.filterwarnings('ignore')

N_BUCKETS = 100   # must match PrepareData (part-000..099)
agg_eng = AggregationEngine(asset=load_asset('aggregation/fe2/trade.json'), table_name='trade')
print('AggregationEngine ready  |  DATA_DIR =', DATA_DIR)
```

```python
BUREAU, SPLIT, VARIANT = 'equifax', 'test', 'new'

proc_dir = os.path.join(DATA_DIR, BUREAU, SPLIT, f'processed_{VARIANT}')
backup   = os.path.join(DATA_DIR, BUREAU, SPLIT, f'processed_{VARIANT}_without_zest_key')
norm_dir = os.path.join(DATA_DIR, BUREAU, SPLIT, f'normalized_{VARIANT}')
assert glob.glob(os.path.join(norm_dir, 'part-*.parquet')), f'no normalized at {norm_dir}'

# 1) back up the keyless processed (first run only)
if glob.glob(os.path.join(backup, 'part-*.parquet')):
    print('backup already exists ->', backup)
elif os.path.isdir(proc_dir):
    shutil.move(proc_dir, backup); print('backed up keyless ->', backup)
os.makedirs(proc_dir, exist_ok=True)
for old in glob.glob(os.path.join(proc_dir, 'part-*.parquet')):
    os.remove(old)

# 2) re-run phase 2 from normalized, saving ZEST_KEY (same buckets + aggregator)
df = pd.read_parquet(norm_dir)
df['_bucket'] = (pd.util.hash_pandas_object(df['ZEST_KEY'], index=False) % N_BUCKETS).astype('int16')
print(f'loaded {len(df):,} normalized rows')
total = 0
for b in range(N_BUCKETS):
    sub = df[df['_bucket'] == b].drop(columns='_bucket')
    if not len(sub):
        continue
    processed = agg_eng.transform(sub).reset_index()
    if 'ZEST_KEY' not in processed.columns and 'index' in processed.columns:
        processed = processed.rename(columns={'index': 'ZEST_KEY'})
    processed.to_parquet(os.path.join(proc_dir, f'part-{b:03d}.parquet'), index=False)
    total += len(processed)
    del sub, processed; gc.collect()
del df; gc.collect()

chk = pd.read_parquet(os.path.join(proc_dir, 'part-000.parquet'))
assert 'ZEST_KEY' in chk.columns, 'ZEST_KEY STILL missing!'
print(f'done {BUREAU}/{SPLIT}/{VARIANT}: {total:,} rows, ZEST_KEY present, {chk.shape[1]} cols')
```
