# Build equifax / valid / OLD  (full pipeline, ZEST_KEY index, checks)

`equifax/valid` was only run for NEW -- no `normalized_old`, no `processed_old`. This runs the whole OLD pipeline: phase 1 (`unmapped` -> `MapperV2` -> `PreprocessorV2` with `missing_data_chars` stripped -> `normalized_old`), then phase 2 (aggregate -> `processed_old` with **ZEST_KEY as the index**), then checks. Run in the model-engine kernel.

```python
import os, sys, gc, glob, warnings
from pathlib import Path
import pandas as pd

for cand in [os.path.abspath(os.path.join(os.getcwd(), '..')), os.getcwd(),
             '/home/jag/payment-processor-research']:
    if os.path.exists(os.path.join(cand, 'configs.py')):
        sys.path.insert(0, cand); break
from configs import EQUIFAX, unmapped_dir, normalized_dir_old, processed_dir_old

from model_engine.assets.utils import load_asset
from model_engine.feature_engine_V2.listed_objects_engines import MapperV2, PreprocessorV2
from model_engine.feature_engine_V2.feature_engine import AggregationEngine
warnings.filterwarnings('ignore')

BUREAU, SPLIT, N_BUCKETS = EQUIFAX, 'valid', 100

asset = load_asset('equifax/cms_6/fe2/trade.json')
for s in asset['preprocess']:
    if s['type'] == 'PaymentPatternsAggregatorV2':
        removed = s['params'].pop('missing_data_chars', None)
        if removed is not None:
            print('stripped missing_data_chars =', removed, '(OLD aggregator)')
mapper       = MapperV2(api=asset['mapping'])
preprocessor = PreprocessorV2(api=asset['preprocess'])
agg_eng      = AggregationEngine(asset=load_asset('aggregation/fe2/trade.json'), table_name='trade')
print('ready: Mapper + Preprocessor + Aggregator')
```

```python
# ---- PHASE 1: unmapped -> normalized_old ----
in_dir   = Path(unmapped_dir(BUREAU, SPLIT))
norm_dir = Path(normalized_dir_old(BUREAU, SPLIT))
parts = sorted(in_dir.glob('part-*.parquet'))
assert parts, f'no unmapped chunks at {in_dir}'
norm_dir.mkdir(parents=True, exist_ok=True)
for f in norm_dir.glob('part-*.parquet'): f.unlink()
print(f'phase 1: {len(parts)} chunks -> {norm_dir}')
for i, in_path in enumerate(parts):
    chunk      = pd.read_parquet(in_path)
    mapped     = mapper.transform(chunk)
    normalized = preprocessor.transform(mapped)
    normalized.to_parquet(norm_dir / f'part-{i:05d}.parquet', index=False)
    if (i + 1) % 50 == 0 or i + 1 == len(parts): print(f'  saved {i + 1}/{len(parts)}')
    del chunk, mapped, normalized; gc.collect()
print('phase 1 done')
```

```python
# ---- PHASE 2: normalized_old -> processed_old (ZEST_KEY as index) ----
proc_dir = Path(processed_dir_old(BUREAU, SPLIT))
proc_dir.mkdir(parents=True, exist_ok=True)
for f in proc_dir.glob('part-*.parquet'): f.unlink()
df = pd.read_parquet(norm_dir)
print(f'phase 2: loaded {len(df):,} normalized rows -> {proc_dir}')
df['_bucket'] = (pd.util.hash_pandas_object(df['ZEST_KEY'], index=False) % N_BUCKETS).astype('int16')
total = 0
for b in range(N_BUCKETS):
    sub = df[df['_bucket'] == b].drop(columns='_bucket')
    if not len(sub): continue
    processed = agg_eng.transform(sub)
    processed.index.name = 'ZEST_KEY'
    processed.to_parquet(proc_dir / f'part-{b:03d}.parquet', index=True)   # index=True
    total += len(processed)
    if (b + 1) % 25 == 0: print(f'  bucket {b + 1}/{N_BUCKETS}, {total:,} ZEST_KEYs')
    del sub, processed; gc.collect()
del df; gc.collect()
print(f'phase 2 done: {total:,} ZEST_KEYs')
```

```python
# ---- CHECKS ----
norm = pd.read_parquet(normalized_dir_old(BUREAU, SPLIT), columns=['ZEST_KEY'])
proc = pd.read_parquet(processed_dir_old(BUREAU, SPLIT))
assert proc.index.name == 'ZEST_KEY', f'ZEST_KEY not the index! got {proc.index.name!r}'
print(f'normalized_old rows / unique ZEST : {len(norm):,} / {norm["ZEST_KEY"].nunique():,}')
print(f'processed_old  rows / unique ZEST : {len(proc):,} / {proc.index.nunique():,}')
print(f'processed_old feature cols        : {proc.shape[1]:,}')
assert len(proc) == proc.index.nunique(), 'duplicate ZEST_KEYs in processed_old!'
assert len(proc) == norm['ZEST_KEY'].nunique(), 'rows != unique normalized ZEST_KEYs!'
print('\nALL CHECKS PASSED -- equifax/valid/old built with ZEST_KEY index')
```
